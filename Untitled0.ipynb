### Cómo instalar bibliotecas de Python en Colab

Para instalar bibliotecas de Python en Google Colab, puedes usar el administrador de paquetes `pip`. Simplemente añade un signo de exclamación `!` antes del comando `pip` en una celda de código.

Por ejemplo, para instalar la biblioteca `pandas`, usarías el siguiente comando:

```python
!pip install pandas
```

Después de instalarla, puedes importar la biblioteca y usarla en tu código.

```python
import pandas as pd

print("La biblioteca pandas ha sido importada exitosamente.")
```
